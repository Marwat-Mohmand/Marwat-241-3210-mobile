```
int width = 10;
int height = 5;
char symbol = '*';
bool isFilled = true;

for (int r = 0; r < height; r++)
{
    for (int c = 0; c < width; c++)
    {
        if (r == 0 || r == height - 1 || c == 0 || c == width - 1)
        {
            Console.Write(symbol);
        }
        else
        {
            Console.Write(' ');
        }
    }

    Console.WriteLine();
}
```

```
string[] GetRectangle(
    int width,
    int height,
    char symbol = '*',
    bool isFilled = true,
    char symbolClear = ' ')
{
    List<string> result = new();

    for (int r = 0; r < height; r++)
    {
        if (isFilled || r == 0 || r == height - 1)
            result.Add(new string(symbol, width));
        else
            result.Add(symbol + new string(symbolClear, width - 2) + symbol);
    }

    return result.ToArray();
}

var shape = GetRectangle(25, 4);
Console.WriteLine(string.Join(Environment.NewLine, shape));
```

```
var shape = GetRectangle(
    20,
    4,
    symbol: '@',
    isFilled: false,
    symbolClear: '-'
);

Console.WriteLine(string.Join(Environment.NewLine, shape));
```
