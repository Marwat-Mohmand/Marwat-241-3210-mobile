```
using System;
using System.Collections.Generic;
using System.Linq;

enum CardSuit
{
    Clubs, Diamonds, Hearts, Spades
}

enum CardRank
{
    Ace = 1, Two, Three, Four, Five, Six, Seven,
    Eight, Nine, Ten, Jack, Queen, King
}

Console.WriteLine("Part 1 ready: suits and ranks created.");
```

```
class Card
{
    public CardSuit Suit { get; }
    public CardRank Rank { get; }

    public Card(CardSuit suit, CardRank rank)
    {
        Suit = suit;
        Rank = rank;
    }

    public override string ToString()
    {
        string symbol = Suit switch
        {
            CardSuit.Clubs => "♣",
            CardSuit.Diamonds => "♦",
            CardSuit.Hearts => "♥",
            CardSuit.Spades => "♠",
            _ => "?"
        };

        string name = Rank switch
        {
            CardRank.Ace => "A",
            CardRank.Jack => "J",
            CardRank.Queen => "Q",
            CardRank.King => "K",
            _ => ((int)Rank).ToString()
        };

        return $"{name}{symbol}";
    }
}

Console.WriteLine(new Card(CardSuit.Hearts, CardRank.Queen));
```

```
class Deck
{
    private readonly List<Card> cards = new();
    private readonly Random random = new();

    public int Count => cards.Count;

    public Deck()
    {
        Reset();
    }

    public void Reset()
    {
        cards.Clear();

        foreach (CardSuit suit in Enum.GetValues<CardSuit>())
        {
            foreach (CardRank rank in Enum.GetValues<CardRank>())
            {
                cards.Add(new Card(suit, rank));
            }
        }
    }

    public void Shuffle()
    {
        for (int i = cards.Count - 1; i > 0; i--)
        {
            int j = random.Next(i + 1);
            (cards[i], cards[j]) = (cards[j], cards[i]);
        }
    }

    public Card Draw()
    {
        if (cards.Count == 0)
            throw new InvalidOperationException("The deck is empty.");

        Card card = cards[^1];
        cards.RemoveAt(cards.Count - 1);
        return card;
    }
}

Console.WriteLine("Deck class ready.");
```

```
var deck = new Deck();
Console.WriteLine($"New deck: {deck.Count} cards");

deck.Shuffle();
Console.WriteLine("Five cards drawn:");

for (int i = 0; i < 5; i++)
{
    Console.WriteLine(deck.Draw());
}

Console.WriteLine($"Cards remaining: {deck.Count}");

deck.Reset();
Console.WriteLine($"After reset: {deck.Count} cards");
```

```
var testDeck = new Deck();
var allCards = new List<Card>();

while (testDeck.Count > 0)
{
    allCards.Add(testDeck.Draw());
}

int uniqueCards = allCards
    .Select(card => card.ToString())
    .Distinct()
    .Count();

Console.WriteLine($"Cards drawn: {allCards.Count}");
Console.WriteLine($"Unique cards: {uniqueCards}");
Console.WriteLine($"Deck empty: {testDeck.Count == 0}");

try
{
    testDeck.Draw();
}
catch (InvalidOperationException ex)
{
    Console.WriteLine(ex.Message);
}
```
