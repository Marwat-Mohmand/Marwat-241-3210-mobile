```
int Sum1(int a, int b)
{
    return a + b;
}

Console.WriteLine(Sum1(1, 2));
Console.WriteLine(Sum1(10, 20));
```

```
void PrintParams2(params int[] values)
{
    Console.WriteLine(string.Join(" | ", values));
}

PrintParams2(new int[] { 1, 2, 3, 4 });
PrintParams2([1, 2, 3, 4]);
PrintParams2(1, 2, 3, 4);
```

```
int Add(int a, int b) => a + b;

string Add(string a, string b) => a + b;

int Add(params int[] numbers)
{
    int sum = 0;

    foreach (var number in numbers)
        sum += number;

    return sum;
}

Console.WriteLine(Add(1, 2));
Console.WriteLine(Add(1, 2, 3));
Console.WriteLine(Add("1", "2"));
```

```
static void Print(this string source)
{
    Console.WriteLine(source);
}

static void Print(this string source, string border)
{
    Console.WriteLine($"{border}{source}{border}");
}

static void PrintUpper(this string source)
{
    Console.WriteLine(source.ToUpper());
}

string st = "Строка";

st.Print();
st.Print("*");
st.Print("##");
st.PrintUpper();

string s2 = "adfafqert43245t";
s2.Print();
s2.PrintUpper();
```
