```
string[] GetRhombus(
    int size,
    char symbol = '*',
    bool isFilled = true,
    char symbolClear = ' ')
{
    List<string> result = new();

    for (int r = 0; r < size; r++)
    {
        int spaces = Math.Abs(size / 2 - r);
        int width = size - 2 * spaces;

        string line;

        if (isFilled || width <= 2)
        {
            line = new string(symbol, width);
        }
        else
        {
            line = symbol
                + new string(symbolClear, width - 2)
                + symbol;
        }

        result.Add(new string(' ', spaces) + line);
    }

    return result.ToArray();
}
var shape = GetRhombus(
    9,
    symbol: '@',
    isFilled: false,
    symbolClear: '-'
);

Console.WriteLine(string.Join(Environment.NewLine, shape));
```

```
string[] GetTriangle(
    int height,
    char symbol = '*',
    bool isFilled = true,
    char symbolClear = ' ')
{
    List<string> result = new();

    for (int r = 0; r < height; r++)
    {
        int width = r * 2 + 1;
        int spaces = height - r - 1;

        string line;

        if (isFilled || r == 0 || r == height - 1)
        {
            line = new string(symbol, width);
        }
        else
        {
            line = symbol
                + new string(symbolClear, width - 2)
                + symbol;
        }

        result.Add(new string(' ', spaces) + line);
    }

    return result.ToArray();
}
var shape = GetTriangle(
    6,
    symbol: '#',
    isFilled: false,
    symbolClear: '.'
);

Console.WriteLine(string.Join(Environment.NewLine, shape));
```

```
string[] GetSmiley()
{
    return new string[]
    {
        "  *****  ",
        " *     * ",
        "*  o o  *",
        "*       *",
        "* \\___/ *",
        " *     * ",
        "  *****  "
    };
}
var shape = GetSmiley();

Console.WriteLine(string.Join(Environment.NewLine, shape));
```

```
string[] GetSmiley(char border = '*', char eye = 'o')
{
    return new string[]
    {
        $"  {new string(border, 5)}  ",
        $" {border}     {border} ",
        $"{border}  {eye} {eye}  {border}",
        $"{border}       {border}",
        $"{border} \\___/ {border}",
        $" {border}     {border} ",
        $"  {new string(border, 5)}  "
    };
}
var shape = GetSmiley(
    border: '#',
    eye: '@'
);

Console.WriteLine(string.Join(Environment.NewLine, shape));
```

```
string[] GetStripedRectangle(
    int width,
    int height,
    char symbol1 = '*',
    char symbol2 = '#')
{
    List<string> result = new();

    for (int r = 0; r < height; r++)
    {
        string line = "";

        for (int c = 0; c < width; c++)
        {
            if ((r + c) % 2 == 0)
                line += symbol1;
            else
                line += symbol2;
        }

        result.Add(line);
    }

    return result.ToArray();
}
var shape = GetStripedRectangle(
    20,
    6,
    symbol1: '@',
    symbol2: '#'
);

Console.WriteLine(string.Join(Environment.NewLine, shape));
```
